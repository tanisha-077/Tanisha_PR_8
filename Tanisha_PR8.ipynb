```python
import numpy as np
```

```python
class DataAnalytics:
    def __init__(self, array=None):
        self.__array = array

    @classmethod
    def create(cls, array):
        return cls(array)

    @staticmethod
    def display(array):
        print(array)

    def create_array(self):
        print("\n1. 1D Array")
        print("2. 2D Array")
        print("3. 3D Array")

        choice = int(input("Enter your choice: "))

        if choice == 1:
            n = int(input("Enter number of elements: "))
            values = list(map(int, input("Enter elements: ").split()))

            if len(values) != n:
                print("Invalid number of elements!")
                return

            self.__array = np.array(values)

        elif choice == 2:
            rows = int(input("Enter number of rows: "))
            cols = int(input("Enter number of columns: "))
            values = list(map(int, input("Enter elements: ").split()))

          

            self.__array = np.array(values).reshape(rows, cols)

        elif choice == 3:
            x = int(input("Enter number of blocks: "))
            y = int(input("Enter number of rows: "))
            z = int(input("Enter number of columns: "))
            values = list(map(int, input("Enter elements: ").split()))

          
            self.__array = np.array(values).reshape(x, y, z)

        else:
            print("Invalid choice!")
            return

        print("\nArray created successfully:")
        print(self.__array)

        self.index_slice()
      

    def index_slice(self):
        if self.__array is None:
            print("Please create an array first!")
            return

        print("\n1. Indexing")
        print("2. Slicing")
        print("3. Go Back")

        choice = input("Enter your choice: ")

        if choice == "1":
            try:
                index = tuple(map(
                    int,
                    input("Enter index (e.g. 0 or 0,1): ").split(",")
                ))
                print("Element:", self.__array[index])
            except (IndexError, ValueError):
                print("Invalid index!")

        elif choice == "2":
            if self.__array.ndim == 2:
                r = input("Enter row range (start:end): ").split(":")
                c = input("Enter column range (start:end): ").split(":")
                print("Sliced Array:")
                print(self.__array[int(r[0]):int(r[1]),
                                   int(c[0]):int(c[1])])
            else:
                start = int(input("Enter start index: "))
                end = int(input("Enter end index: "))
                print("Sliced Array:")
                print(self.__array[start:end])

    def mathematical_operations(self):
        if self.__array is None:
            print("Please create an array first!")
            return

        print("\n1. Addition")
        print("2. Subtraction")
        print("3. Multiplication")
        print("4. Division")
        print("5. Dot Product")
        print("6. Matrix Multiplication")

        choice = input("Enter your choice: ")

        values = list(map(
            float,
            input("Enter elements of second array: ").split()
        ))

        if len(values) != self.__array.size:
            print("Number of elements must be", self.__array.size)
            return

        second = np.array(values).reshape(self.__array.shape)

        print("\nOriginal Array:")
        print(self.__array)
        print("\nSecond Array:")
        print(second)

        if choice == "1":
            result = self.__array + second
        elif choice == "2":
            result = self.__array - second
        elif choice == "3":
            result = self.__array * second
        elif choice == "4":
            if np.any(second == 0):
                print("Division by zero is not allowed!")
                return
            result = self.__array / second
        elif choice == "5":
            result = np.dot(self.__array, second)
        elif choice == "6":
            if self.__array.ndim != 2:
                print("Matrix multiplication requires 2D arrays!")
                return
            if self.__array.shape[1] != second.shape[0]:
                print("Matrix dimensions are incompatible!")
                return
            result = np.matmul(self.__array, second)
        else:
            print("Invalid choice!")
            return

        print("\nResult:")
        print(result)

    def combine_split(self):
        if self.__array is None:
            print("Please create an array first!")
            return

        print("\n1. Combine Arrays")
        print("2. Split Array")

        choice = input("Enter your choice: ")

        if choice == "1":
            values = list(map(
                int,
                input("Enter elements for second array: ").split()
            ))

            if len(values) != self.__array.size:
                print("Enter exactly", self.__array.size, "elements!")
                return

            second = np.array(values).reshape(self.__array.shape)

            if self.__array.ndim == 1:
                result = np.concatenate((self.__array, second))
            else:
                result = np.concatenate((self.__array, second), axis=0)

            print("Combined Array:")
            print(result)

        elif choice == "2":
            parts = int(input("Enter number of parts: "))

            if parts <= 0 or self.__array.size % parts != 0:
                print("Array cannot be split into equal parts!")
                return

            result = np.array_split(self.__array, parts, axis=0)

            print("Split Arrays:")
            for part in result:
                print(part)

    def search_sort_filter(self):
        if self.__array is None:
            print("Please create an array first!")
            return

        print("\n1. Search a Value")
        print("2. Sort Array")
        print("3. Filter Values")

        choice = input("Enter your choice: ")

        if choice == "1":
            value = float(input("Enter value to search: "))
            positions = np.argwhere(self.__array == value)

            if len(positions) > 0:
                print("Value found at indices:")
                print(positions)
            else:
                print("Value not found!")

        elif choice == "2":
            order = input("Ascending or descending (a/d): ").lower()
            result = np.sort(self.__array, axis=None)

            if order == "d":
                result = result[::-1]

            print("Sorted Array:")
            print(result)

        elif choice == "3":
            value = float(input("Show elements greater than: "))
            print("Filtered Values:")
            print(self.__array[self.__array > value])

    def statistics(self):
        if self.__array is None:
            print("Please create an array first!")
            return

        print("\n1. Sum")
        print("2. Mean")
        print("3. Median")
        print("4. Standard Deviation")
        print("5. Variance")
        print("6. Minimum and Maximum")
        print("7. Percentile")

        choice = input("Enter your choice: ")

        if choice == "1":
            result = np.sum(self.__array)
        elif choice == "2":
            result = np.mean(self.__array)
        elif choice == "3":
            result = np.median(self.__array)
        elif choice == "4":
            result = np.std(self.__array)
        elif choice == "5":
            result = np.var(self.__array)
        elif choice == "6":
            print("Minimum:", np.min(self.__array))
            print("Maximum:", np.max(self.__array))
            return
        elif choice == "7":
            p = float(input("Enter percentile (0-100): "))
            if not 0 <= p <= 100:
                print("Percentile must be between 0 and 100!")
                return
            result = np.percentile(self.__array, p)
        else:
            print("Invalid choice!")
            return

        print("Result:", result)

    def correlation(self):
        if self.__array is None:
            print("Please create an array first!")
            return

        if self.__array.ndim != 1:
            print("Use a 1D array for this operation!")
            return

        values = list(map(
            float,
            input("Enter elements of second array: ").split()
        ))

        if len(values) != self.__array.size:
            print("Both arrays must have the same size!")
            return

        second = np.array(values)

        if np.std(self.__array) == 0 or np.std(second) == 0:
            print("Correlation is undefined for a constant array!")
            return

        print("Correlation Coefficient:",
              np.corrcoef(self.__array, second)[0, 1])


def main():
    analyzer = DataAnalytics()

    while True:
        print("\n       NumPy Analyzer")
        print("1. Create a NumPy Array")
        print("2. Mathematical Operations")
        print("3. Combine or Split Arrays")
        print("4. Search, Sort, or Filter Arrays")
        print("5. Aggregates and Statistics")
        print("6. Correlation Coefficient")
        print("7. Exit")

        choice = input("Enter your choice: ")

        if choice == "1":
            analyzer.create_array()
        elif choice == "2":
            analyzer.mathematical_operations()
        elif choice == "3":
            analyzer.combine_split()
        elif choice == "4":
            analyzer.search_sort_filter()
        elif choice == "5":
            analyzer.statistics()
        elif choice == "6":
            analyzer.correlation()
        elif choice == "7":
            print("Thank you for using the NumPy Analyzer!")
            print("Goodbye!")
            break
        else:
            print("Invalid choice! Try again.")


if __name__ == "__main__":
    main()
    
```
